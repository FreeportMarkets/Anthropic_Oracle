# Anthropic pre-IPO perpetual history

Reproducible analysis of completed UTC daily candles through 2026-08-02. The production oracle uses live impact prices; this notebook is a separate historical sanity check.

```python
import importlib
import sys
from datetime import date
from pathlib import Path

import pandas as pd
from IPython.display import Image, display

ROOT = Path.cwd()
if not (ROOT / "research").exists():
    ROOT = ROOT.parent
sys.path.insert(0, str(ROOT))
history = importlib.import_module("research.build_history")
daily = pd.read_csv(ROOT / "research/data/anthropic_perp_daily.csv")
wide = history.make_wide(daily)
summary = history.compute_summary(wide, date.fromisoformat("2026-08-03"))
wide.tail()
```

```python
pd.Series(summary).drop("daily_return_correlations").to_frame("value")
```

```python
pd.DataFrame(summary["daily_return_correlations"]).round(3)
```

```python
charts = ROOT / "research/charts"
history.create_charts(wide, charts)
for name in [
    "implied-valuation-history",
    "daily-volatility",
    "liquidity-and-dispersion",
]:
    display(Image(filename=charts / f"{name}.png"))
```

## Interpretation limits

Daily close medians demonstrate cross-venue price discovery and are not a substitute for the live 30-second impact-price oracle. Reported volume may include market-maker and other non-directional turnover.
